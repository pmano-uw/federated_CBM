```python
import pandas as pd
import matplotlib.pyplot as plt
import random
from pathlib import Path
from matplotlib.lines import Line2D
from utilities import extract_prefix_and_number

# --- Style setup ---
plt.rcParams.update({
    'font.family': 'sans-serif',
    'font.sans-serif': ['Myriad Pro', 'Fira Sans', 'Source Sans Pro', 'Helvetica Neue'],
    'font.size': 14,
    'axes.titlesize': 15,
    'axes.labelsize': 14,
    'xtick.labelsize': 12,
    'ytick.labelsize': 12,
    'legend.fontsize': 12,
    'figure.titlesize': 15,
    'axes.spines.top': False,
    'axes.spines.right': False,
    'axes.linewidth': 0.8,
    'xtick.direction': 'in',
    'ytick.direction': 'in',
    'xtick.major.width': 0.8,
    'ytick.major.width': 0.8,
    'legend.framealpha': 0.9,
    'legend.edgecolor': '0.8',
    'legend.handlelength': 2.5,
    'pdf.fonttype': 42,
    'ps.fonttype': 42,
})

COLORS = ['#2166AC', '#4DAC26', '#B2182B', '#D6604D', '#762A83', '#1B7837', '#E08214', '#8073AC', '#01665E', '#4D4D4D']

# --- 1. Data Loading ---
dataset_name, dataset_num = extract_prefix_and_number("nasa1")
parent_path = Path.cwd().parent / Path("dataset") / Path(dataset_name)
file_path = Path(f"train_FD00{int(dataset_num)}.txt")

df = pd.read_csv(parent_path / file_path, sep=" ", header=None)
df.drop(columns=[26,27], inplace=True)

columns = ['unit_number','time_in_cycles','setting_1','setting_2','TRA','T2','T24','T30','T50','P2','P15','P30','Nf',
        'Nc','epr','Ps30','phi','NRf','NRc','BPR','farB','htBleed','Nf_dmd','PCNfR_dmd','W31','W32']
df.columns = columns
df = df[['unit_number', 'time_in_cycles', 'T24']].copy()

# --- 2. Calculate RUL for alignment ---
max_cycles = df.groupby('unit_number')['time_in_cycles'].transform('max')
df['RUL'] = max_cycles - df['time_in_cycles']

# --- 3. Signal Processing (Per Unit) ---
baselines = df[df['time_in_cycles'] <= 10].groupby('unit_number')['T24'].mean()
df['T24_baseline_adj'] = df.apply(lambda row: row['T24'] - baselines[row['unit_number']], axis=1)

span_value = 15
df['T24_smoothed'] = df.groupby('unit_number')['T24_baseline_adj'].transform(lambda x: x.ewm(span=span_value, adjust=False).mean())

# --- NEW: Calculate Thresholds ---
# Extract the smoothed values exactly at failure (RUL == 0) across all units
failure_values = df[df['RUL'] == 0]['T24_smoothed']
mean_threshold = failure_values.mean()
conservative_threshold = failure_values.quantile(0.025)

# --- 4. Plotting ---
all_units = df['unit_number'].unique().tolist()
sampled_units = random.sample(all_units, 10)

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 5.5), sharey=True)

for idx, unit in enumerate(sampled_units):
    unit_data = df[df['unit_number'] == unit]
    
    # Left Plot: Raw Data
    ax1.plot(unit_data['RUL'], unit_data['T24_baseline_adj'], label=f"unit_{unit}", color=COLORS[idx], lw=1.5, alpha=0.7)
    
    # Right Plot: Smoothed Data
    ax2.plot(unit_data['RUL'], unit_data['T24_smoothed'], label=f"unit_{unit}", color=COLORS[idx], lw=2.0, alpha=0.9)

# Format both axes
for ax in [ax1, ax2]:
    ax.invert_xaxis() 
    ax.set_xlabel("Remaining Useful Life (Cycles to Failure)")
    ax.grid(True, which='major', linestyle='--', linewidth=0.5, alpha=0.4)
    ax.grid(True, which='minor', linestyle=':', linewidth=0.3, alpha=0.25)
    
    # --- NEW: Draw the horizontal threshold lines ---
    # zorder=10 ensures the threshold lines render on top of the signal trajectories
    ax.axhline(y=mean_threshold, color='black', linestyle='--', linewidth=2.5, alpha=0.8, zorder=10)
    ax.axhline(y=conservative_threshold, color='dimgrey', linestyle=':', linewidth=2.5, alpha=0.8, zorder=10)

# Specific formatting for Left Plot
ax1.set_ylabel(r"$\Delta$ T24 (Baseline Adjusted)")
ax1.set_title("(a) Raw Degradation Signal")

# Specific formatting for Right Plot
ax2.set_title(f"(b) EWMA Smoothed Signal (span={span_value})")

# --- NEW: Append threshold lines to the legend ---
handles, labels = ax2.get_legend_handles_labels()
# Add proxy artists for the threshold lines
handles.extend([
    Line2D([0], [0], color='black', linestyle='--', lw=2.5),
    Line2D([0], [0], color='dimgrey', linestyle=':', lw=2.5)
])
labels.extend([
    f'Mean Failure ({mean_threshold:.2f})', 
    f'Conservative ({conservative_threshold:.2f})'
])

ax2.legend(handles=handles, labels=labels, loc='center left', bbox_to_anchor=(1.02, 0.5), frameon=True, ncol=1)

fig.tight_layout()
plt.show()
```
